```python
import numpy as np

arr = np.array([1, 3, 5, 7])
print('numpy_array:', arr)

arr.shape
```

```python
arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9]])
print('numpy_array2:', arr2)

arr2.shape
```

```python
#dtype()
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape()
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print("shaped_array:", b)
```

```python
# astype()
a = np.array([1, 2, 3])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeros(), np.ones() => dummy data
z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros:", z)
print("ones:", o)
```

```python
# ndim      nechta olchov borligini aniqlab beradi   1D
a = np.array([1, 2])
print("ndim:", a.ndim)
```

```python
# ndim      nechta olchov borligini aniqlab beradi 2D mi
a = np.array([[1, 2] , [3, 4]])
print("ndim:", a.ndim)
```

```python
# arange()            bu test dani hosil qilish un kerak malumot kam bolsa
a = np.arange(1, 7)
print("arange:", a)
```

```python
#sum(), mean(), max(), min()

a = np.array([1, 2, 3, 4])

print("sum:", a.sum()) #qoshish
print("mean:", a.mean()) # sonlarning ortachasini aniqlab beradi
print("max:", a.max()) # arraylarning eng kotta raqamni aniqlab beradi
print("min:", a.min()) # arraylarning eng kichik raqamni aniqlab beradi
```

```python
# flatten()   np da rasmni 1D korinishga keltirish un ishlatiladi yani 2D variantni 1D ga aylantirib berad

a = np.array([[1, 2], [3, 4]])
print('flatten:', a.flatten())
```

```python
# T (Transpose)  array shaklini ozgartirishda, malumotlarni qayta tashkil qilishda
#orin almashadi 1->3, 2->4

a = np.array([[1, 2], [3, 4]])
print("transpose:", a.T)
```

```python
# dot() matrix ni kopaytirishda ishlatiladi

a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("dot/multiplication:", a.dot(b)) # a dot methodan use qilib b ga kopaytiramiz
```

```python
# random.rans()
a = np.random.rand(2, 3) #2 qator- 3ta ustun/.    => 0.0 and 1.0/ 0.0 included / 1.0 excluded bu raqmni korsatmaydi
print("random rand:", a)
```
